```python
df['Date'] = pd.to_datetime(df['Date'])

print("First Date:", df['Date'].min())
print("Last Date:", df['Date'].max())
print("Number of Countries:", df['Country/Region'].nunique())
```

# The dataset covers the period from 22 January 2020 to 24 March 2020. It contains weather records from 163 countries.

```python
weather_columns = [
    'Country/Region', 'Province/State', 'Date',
    'temp', 'min', 'max', 'rh', 'wdsp', 'prcp', 'fog'
]

weather_df = df[weather_columns].copy()

weather_df.head()
```

```python
print("Missing Values in Each Column:")
print(weather_df.isnull().sum())
```

```python
weather_df['min'] = weather_df['min'].fillna(weather_df['min'].median())

weather_df['max'] = weather_df['max'].fillna(weather_df['max'].median())

weather_df['rh'] = weather_df['rh'].fillna(weather_df['rh'].median())

print(weather_df.isnull().sum())
```

# Missing Values Analysis

Missing values were found in the Province/State, minimum temperature, maximum temperature, and relative humidity columns.

The missing values in minimum temperature, maximum temperature, and relative humidity were filled using the median method.

The missing Province/State values were kept unchanged because some countries may not have province-level information available.

After handling the missing values, all weather-related columns contain zero missing values.

```python
print("Duplicate Rows:", weather_df.duplicated().sum())
```

# Duplicate Values Analysis

The dataset was checked for duplicate rows using the duplicated() function.

The result showed 0 duplicate rows. This means no completely repeated records were found, so no duplicate rows needed to be removed.

The dataset will now be checked to confirm that each location has only one record for each date.

```python
print(
    "Duplicate Country-Date Records:",
    weather_df.duplicated(subset=['Country/Region', 'Province/State', 'Date']).sum()
)
```

# Checking Repeated Records by Country and Date

The dataset was checked for repeated records using Country/Region, Province/State, and Date.

The result showed 0 duplicate records. This means no repeated records were found for the same country, province, and date. Therefore, no additional duplicate removal was required.

```python
import matplotlib.pyplot as plt
import seaborn as sns

weather_columns = ['temp', 'prcp', 'wdsp']

for col in weather_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=weather_df[col])
    plt.title(f'Boxplot of {col}')
    plt.xlabel(col)
    plt.show()
```

```python
for col in ['temp', 'prcp', 'wdsp']:
    plt.figure(figsize=(8, 4))
    sns.histplot(weather_df[col], kde=True)
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
for col in ['temp', 'prcp', 'wdsp']:
    Q1 = weather_df[col].quantile(0.25)
    Q3 = weather_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = weather_df[
        (weather_df[col] < lower_limit) |
        (weather_df[col] > upper_limit)
    ]

    print(f"\n{col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))
```

# Outlier Detection Using IQR Method

The Interquartile Range (IQR) method was used to identify unusual values in temperature, rainfall, and wind speed.

The analysis detected 36 outliers in temperature, 3,818 outliers in rainfall, and 908 outliers in wind speed.

Rainfall had the highest number of outliers. These values may represent genuine weather variations rather than errors. Therefore, the outliers were retained for further analysis instead of being removed automatically.

```python
for col in ['temp', 'prcp', 'wdsp']:
    print("\n", col)
    print("Minimum:", weather_df[col].min())
    print("Maximum:", weather_df[col].max())
```

```python
import numpy as np

weather_df.loc[weather_df['prcp'] == 99.99, 'prcp'] = np.nan
weather_df.loc[weather_df['wdsp'] == 999.9, 'wdsp'] = np.nan

print(weather_df[['prcp', 'wdsp']].isnull().sum())
```

# Checking Weather Data Extremes

During the extreme-value analysis, unusually high values were found in rainfall and wind speed.

The value 99.99 in rainfall and 999.9 in wind speed were identified as missing-data codes rather than genuine weather measurements. These values were replaced with missing values (NaN).

A total of 1,389 missing values were identified in rainfall and 343 in wind speed after this correction. Other unusual weather values were retained for further analysis.

```python
weather_df['prcp'] = weather_df['prcp'].fillna(weather_df['prcp'].median())

weather_df['wdsp'] = weather_df['wdsp'].fillna(weather_df['wdsp'].median())

print(weather_df.isnull().sum())
```

# Handling Missing Values After Outlier Checking

After identifying the missing-data codes in rainfall and wind speed, these values were replaced with the median of their respective columns.

The final check shows that all weather-related columns contain zero missing values. The missing values in Province/State were kept because province-level information is not available for every location.

The dataset is now ready for further analysis.

```python
# Check temperature outliers
temp_outliers = weather_df[
    (weather_df['temp'] < -18.2) |
    (weather_df['temp'] > 133.8)
]

# Check rainfall outliers
prcp_outliers = weather_df[
    (weather_df['prcp'] < -0.045) |
    (weather_df['prcp'] > 0.075)
]

# Check wind speed outliers
wdsp_outliers = weather_df[
    (weather_df['wdsp'] < -4.3) |
    (weather_df['wdsp'] > 16.5)
]

print("Temperature outliers:", len(temp_outliers))
print("Rainfall outliers:", len(prcp_outliers))
print("Wind speed outliers:", len(wdsp_outliers))
```

# Outlier Analysis and Decision

The IQR method was used to identify outliers in temperature, rainfall, and wind speed.

The analysis identified 36 temperature outliers, 2,429 rainfall outliers, and 565 wind speed outliers.

These values were reviewed after checking the minimum and maximum values. Since extreme weather conditions can occur naturally, the outliers were not removed. Keeping them helps preserve important weather information for further analysis.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=weather_df, x='temp', y='rh')

plt.title('Relationship Between Temperature and Humidity')
plt.xlabel('Temperature')
plt.ylabel('Relative Humidity (%)')

plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = weather_df[
    ['temp', 'min', 'max', 'rh', 'wdsp', 'prcp']
].corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Weather Variables Correlation Heatmap')

plt.show()
```

### Correlation Heatmap Analysis

The correlation heatmap displays the relationships between temperature, minimum temperature, maximum temperature, relative humidity, wind speed, and rainfall. It uses colours and numerical values to show how strongly these weather variables are related. Positive values indicate variables moving in the same direction, while negative values indicate variables moving in opposite directions.

```python
daily_temp = weather_df.groupby('Date')['temp'].mean()

plt.figure(figsize=(12, 5))

plt.plot(daily_temp.index, daily_temp.values)

plt.title('Average Temperature Over Time')
plt.xlabel('Date')
plt.ylabel('Average Temperature')

plt.xticks(rotation=45)
plt.show()
```

### Average Temperature Over Time

The line chart shows how the average temperature changes over the selected dates from January to March 2020. It helps identify temperature fluctuations and changes over time across the locations included in the dataset.

```python
country_temp = (
    weather_df.groupby('Country/Region')['temp']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(x=country_temp.values, y=country_temp.index)

plt.title('Top 10 Countries by Average Temperature')
plt.xlabel('Average Temperature')
plt.ylabel('Country')

plt.show()
```

```python
print("Temperature and Humidity Correlation:")
print(weather_df['temp'].corr(weather_df['rh']))
```
